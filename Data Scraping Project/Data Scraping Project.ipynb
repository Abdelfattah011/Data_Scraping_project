### Imports And Options

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import pandas as pd
import time

options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
#options.add_argument("--headless=new")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
```

### Scraping Code

```python
driver = webdriver.Chrome(options=options)
driver.get('https://www.naukrigulf.com/')
list_jobs= []
time.sleep(10)   #We use Sleep to wait for the page to load completely before performing any actions on it. To Avoid any errors.

search_click = driver.find_element(By.XPATH, '//*[@id="ngHomePage"]/div[1]/div[1]/div[1]').click()
search_job = driver.find_element(By.XPATH, '//*[@id="qsbKey"]').send_keys('Data Engineer')
time.sleep(3)
search_click_again = driver.find_element(By.XPATH, '//*[@id="ngQsbForm"]/input').click()
time.sleep(5)
for i in range(3):
 
    
 job_titles = driver.find_elements(By.CLASS_NAME, 'designation-title')
 job_companies = driver.find_elements(By.CLASS_NAME, 'info-org')
 job_locations = driver.find_elements(By.CLASS_NAME, 'info-loc')
 job_experiences = driver.find_elements(By.CLASS_NAME, 'info-exp')
 job_descriptions = driver.find_elements(By.CLASS_NAME, 'description')

 
 for title, company, location, experience, description in zip(job_titles, job_companies, job_locations, job_experiences, job_descriptions):
    list_jobs.append({
        'job_title': title.text,
        'job_company': company.text,
        'job_location': location.text,
        'job_experience': experience.text,
        'job_description': description.text
    })
    time.sleep(1)

                              

 
 

   
 Next_page_click = driver.find_element(By.XPATH, '//*[@id="pagination"]/ul/li[7]/span').click()

 time.sleep(5)
```

### Export the output to a CSV file

```python
df = pd.DataFrame(list_jobs)
df.to_csv('jobs.csv', index=False)    
```
